# Random Forest (Clasificación binaria)
El onjectivo de este punto es entrenar un modelo RandomForestClassifier que prediga si un tumor es maligno/benigno haciendo uso del dataset: sklearn.datasets.load_breast_cancer()

## Paso 1 - Importar librerias

```python
%matplotlib inline
import matplotlib
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
import numpy as np
import seaborn as sns
```

## Paso 2 - Cargar Dataset y la hacer el Análisis Exploratorio de los datos

```python
cancer = load_breast_cancer()
print("Target names:", cancer.target_names)
print(cancer.DESCR)

df_cancer = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df_cancer['target'] = cancer.target
from IPython.display import display

display(df_cancer.head())
df_cancer.info()
print('Dimensiones (filas, columnas):', df_cancer.shape)
print('Tipos de datos por columna:')
print(df_cancer.dtypes)

```

## Paso 3 - Objectivo del modelo
Al entrenar un modelo de regresión lineal se tiene una variable a predecir y las variables predictoras. En este caso, buscar+mos predecir si un tumor es Maligno o es Benigno de acuerdo a la dependencia de la variable a predecir (target) con las demás variables.
