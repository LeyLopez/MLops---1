Dataset: sklearn.datasets.load_breast_cancer()
Objetivo: Entrenar un modelo RandomForestClassifier que prediga si un tumor es maligno/benigno.
Entrega (obligatoria):

## Paso 1 - Importar librerias y cargar dataset
Inicialmente, se realiza la importación de las librerías necesarias para realizar el modelo, se carga el dataset y se muestran sus primeras filas.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report
)
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

#Loading dataset
cancer = load_breast_cancer()
X = cancer.data      #Separación de variables predictoras
y = cancer.target    #Variable a predecir

#Exploring dataset
df = pd.DataFrame(X, columns=cancer.feature_names)
df['target'] = y

print('Shape del dataset: ', df.shape)
df.head()
```

## Paso 2 - Exploración de datos
A continuación, se realiza una exploración de los datos. Se visualizan las estadisticas descriptivas y la información principal del dataset. Luego, se grafíca la distribución de las clases (Benigno y Maligno) dentro del conjunto de datos. Finalmente, se realizar una matriz de correlación entre las variables predictoras.

```python
print('Estadisticas descriptivas del dataset: \n', df.describe())
print('Información general del dataset: \n', df.info())

# Distribución de clases (benigno - Maligno)
sns.countplot(x='target', data=df)
plt.xticks([0,1], cancer.target_names)
plt.title("Distribución de clases (Benigno vs Maligno)")
plt.show()

print("Proporción de clases:\n", df['target'].value_counts(normalize=True))

# Matriz de correlación entre variables predictoras
plt.figure(figsize=(14,10))
corr = df.drop('target', axis=1).corr()
sns.heatmap(corr, cmap="coolwarm", center=0, square=True,
            linewidths=.5, cbar_kws={"shrink": .5})
plt.title("Matriz de correlación entre variables", fontsize=16)
plt.show()
```

## Paso 3 - División de datos de entrenamientos y prueba del modelo
Se realiza la división entre los datos de entrenamiento y de prueba utilizando train_test_split.

```python
from sklearn.model_selection import train_test_split


X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train size:", X_train.shape[0])
print("Test size:", X_test.shape[0])
```

## Paso 4 - Entrenamiento del modelo de RandomForestClassifier
Ahora bien, procedemos al entrenamiento del modelo de Random Forest, con un numero de arboles de 100, sin limite de profundidad y con una semilla de 42 (mayormente utilizada).

```python
clf = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    random_state=42
)

clf.fit(X_train, y_train)
```

## Paso 5 - Evaluación con datos de prueba
Ahora procedemos a hacer la evaluación del modelo con los datos de prueba, mostrando las metricas y la matriz de confusión de los resultados de la evaluación.

```python
y_pred = clf.predict(X_test)
y_proba = clf.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1-score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))

# Matriz de confusión
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=cancer.target_names,
            yticklabels=cancer.target_names)
plt.xlabel("Predicción")
plt.ylabel("Real")
plt.title("Matriz de Confusión")
plt.show()
```

## Paso 6 - Importancia de las variables
En este paso, graficamos el Top 10 de las caracteristicas o variables más importantes durante el entrenamiento del modelo.

```python
importances = pd.Series(clf.feature_importances_, index=cancer.feature_names)
importances.sort_values(ascending=False).head(10).plot(kind="barh")
plt.title("Top 10 Features más importantes")
plt.show()
```

## Paso 7 -  Análisis de los resultados
De acuerdo a lo anterior, podemos afirmar que el modelo de Random Forest Cassifier arroja metricas muy buenas con el dataset.
Podemos observar, que el accuracy y el recall altos reflejan que está detectando correctamente la mayoria de los tumores malignos y benignos. El ROC-AUC nos da a entender que está realizando una correcta separación entre las clases a predecir.

Al graficar las variables más importantes nos podemos dar cuenta que worst area y worst concave points juegan un papel clave en la predicción del modelo.
